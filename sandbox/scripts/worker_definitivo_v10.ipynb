```python
import os
import numpy as np
from kaggle_light import crear_kernel, activar_kernel
import requests
import time

# Configuración del worker
tema = 'lofi jazz con saxofón y lluvia'

# Definición del modelo ACE-Step
from ace_step import load_model
model = load_model('acestep-15-models')

# Función para generar música con saxofón y ambiente de lluvia
def generar_musica_lofi_jazz(saxofon=False, lluvia=True):
    """Genera una pista de lofi jazz con saxofón y efectos de lluvia"""
    params = {
        'tempo': 80,
        'key': 'minor',
        'instrumentos': ['saxofon', 'piano', 'piano_acustico'] if saxofon else ['piano', 'synth'],
        'ambiente': 'lluvia' if lluvia else 'silencioso',
        'duracion_segundos': 240
    }
    return model.generate(**params)

# Punto de entrada principal
if __name__ == '__main__':
    print(f'Worker iniciado para tema: {tema}')
    print("Generando música de lofi jazz con saxofón y lluvia...")
    musica = generar_musica_lofi_jazz(saxofon=True, lluvia=True)
    print(f'Música generada con longitud: {len(musica)}')
```

# Worker de Música Lofi

Este notebook es el núcleo del pipeline de Kaggle para generar música de estilo lofi jazz con saxofón y efectos de lluvia.

## Funcionalidad
- Genera pistas de música utilizando modelos ACE-Step
- Permite configurar presencia de saxofón y efectos de lluvia
- Devuelve una señal de audio o datos estructurados

## Parámetros
- `tema`: Nombre descriptivo de la canción (por ejemplo: "lofi jazz con saxofón y lluvia")
- `saxofon`: Booleano, si True incluye saxofón en la mezcla
- `lluvia`: Booleano, si True añade efectos atmosféricos de lluvia
